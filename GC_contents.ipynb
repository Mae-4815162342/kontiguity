```python
from Bio import SeqIO
import os
import numpy as np
import pandas as pd
```

```python
def get_GC(sequence):
    """Computes GC content of a sequence"""
    return np.sum([1 if s in ["G", "C"] else 0 for s in sequence])/len(sequence)

def get_all_GC(fasta):
    """Computes GC content of every sequence in the fasta file"""
    fasta_sequences = SeqIO.parse(open(fasta),'fasta')
    GC_contents = {}
    for fasta in fasta_sequences:
        name, sequence = fasta.id, str(fasta.seq)
        GC_contents[name] = get_GC(sequence)
    df_GC = pd.DataFrame([
        list(GC_contents.keys()),
        list(GC_contents.values()),
    ]).T
    df_GC.columns = ["Sequence", "GC content"]
    return df_GC
```
