```python
def get_GC(sequence):
    """Computes GC content of a sequence"""
    return np.sum([1 if s in ["G", "C"] else 0 for s in sequence])/len(sequence)

# Palindrome detection from An Algorithm for Detection of Palindrome DNA
# By Brian W. Li - November 5, 2025
# written by ChatGPT

def reverse_complement(S):
    """Return the reverse complement of a DNA sequence."""
    comp = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
    return ''.join(comp.get(b, b) for b in reversed(S))


def score(length):
    """
    Scoring function placeholder.
    Can be replaced by any scoring scheme you want.
    """
    return length

def palindromes(S, n, D = 0.05, X = 20, g = 200, L = 5):
    """
    Detect palindromes in sequence S.
    
    Parameters:
        S : str  - input sequence
        n : int  - length of S
        D : float - maximum mismatch ratio
        X : float - x-drop threshold
        g : int  - maximum gap
        L : int  - minimum palindrome length

    Returns:
        V : list - list of detected palindromes
    """

    # Initialize output
    V = []

    # Reverse complement of S
    Sc = reverse_complement(S)

    index = 0
    while index < n:

        # Furthest-reaching paths on each diagonal
        W = [0 for _ in range(g + 1)]

        # Number of mismatches
        E = 0

        # Highest score
        M = 0

        # Longest path (small value to avoid division by zero)
        l = 0.001

        # Extension phase
        while (E / l) < D:

            for w in range(len(W)):

                # Extend while bases match
                while (index + W[w] < n and
                       index + W[w] < len(Sc) and
                       S[index + W[w]] == Sc[index + W[w]]):

                    W[w] += 1

                # Update max score
                M = max(score(W[w]), M)

                # Update max length
                l = max(W[w], l)

            # One mismatch encountered
            E += 1

            # X-drop or mismatch-ratio stopping condition
            if score(l) < M - X or (M / l) >= D:
                break

        # Check minimum palindrome length
        if l >= L:
            # Palindrome representation (example)
            P = {
                "start": index,
                "length": int(l),
                "sequence": S[index:index + int(l)]
            }
            V.append(P)

            # Skip to the end of this palindrome
            index += int(l)
        else:
            index += 1

    return V
```
