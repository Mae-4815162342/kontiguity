```python
import cooler
import os
import numpy as np
import pandas as pd

def compute_number_of_copies(cool, chromosomes, contig):
    """Estimates the number of copies of a sequence in the genome from Hi-C data (cool file). 
    The following formula is applied:
        # copies = Cov(contig)/Median(Cov(chromosome))
    with Cov(x) = Reads(x)/Size(x) for any sequence x.
    """
    sequences = np.concatenate([chromosomes, np.array([contig])])
    matrix = cool.matrix(balance = False)
    reads = {seq: 0 for seq in sequences}
    # Estimating each sequence number of reads from raw data
    for i in range(len(sequences)):
        for j in range(i, len(sequences)):
            if i == j:
                local_reads = np.nansum(matrix.fetch(sequences[i])[:])
                reads[sequences[i]] += local_reads // 2 # divided by 2 because of the matrix symetry, simplified by ignoring the diagonal
            else:
                local_reads = np.nansum(matrix.fetch(sequences[i], sequences[j])[:])
                reads[sequences[i]] += local_reads
                reads[sequences[j]] += local_reads
    
    # Computing the average cov of the chromosomes
    chrom_coverages = [
        reads[chrom] / cool.chromsizes[chrom] for chrom in chromosomes
    ]
    average_cov = np.nanmedian(chrom_coverages)
    
    # computing contig coverage
    contig_coverage = reads[contig] / cool.chromsizes[contig]
    
    return contig_coverage/average_cov
```

```python
plasmid = "AP023147.1"
library = "Entamoeba_histolytica"
binning = 100000
cool_file = f"/media/sardine/data_2/Eukaroytic_plasmids_manual_research/mcools/{library}.mcool"
plot_outpath = f"/media/sardine/data_2/Eukaroytic_plasmids_manual_research/contact_plots/{library}"
if not os.path.exists(plot_outpath):
    os.mkdir(plot_outpath)
cool = cooler.Cooler(f"{cool_file}::resolutions/{binning}")
chromosomes = np.array([ chrom for chrom in list(cool.chromsizes.keys()) if chrom[:3] == "ENA" or chrom[:3] == "NC_" or (chrom[:3] != "ERR" and chrom[0] != "k" and chrom[:3] != "tig" and chrom != plasmid)])

for i in range(len(chromosomes)):
    chrom_tmp = chromosomes[1:] if i == 0 else np.concatenate([chromosomes[:i-1], chromosomes[i + 1:]]) if i < len(chromosomes) else chromsomes[:-1]
    nb_copies = compute_number_of_copies(cool, chrom_tmp, chromosomes[i])
    print(f"Chromosome {chromosomes[i]}: {nb_copies} copies.")

nb_copies = compute_number_of_copies(cool, chromosomes, plasmid)
print(f"Contig {plasmid}: {nb_copies} copies.")
```

## Looking for high copy number in contigs

```python
def coverage(matrix, sequence, chromosomes):
    """Return the number of reads"""
    sequences = chromosomes
    nb_reads = 0
    for i in range(len(sequences)):
        local_reads = np.nansum(matrix.fetch(sequence)[:]) if sequences[i] == sequence else np.nansum(matrix.fetch(sequences[i], sequence)[:])
        nb_reads += local_reads if sequences[i] == sequence else local_reads // 2 # divided by 2 because of the matrix symetry, simplified by ignoring the diagonal
    return nb_reads
    
def compute_all_contigs_copy_number(cool, chromosomes, ploidy = 1):
    """Estimates the number of copies of all the contigs in the genome from Hi-C data (cool file). 
    The following formula is applied:
        # copies = Cov(contig)/Median(Cov(chromosome))
    with Cov(x) = Reads(x)/Ploidy*Size(x) for any sequence x, 
    with the Ploidy = 2 for Diploid genomes, and 1 for Haploid genomes.
    """
    matrix = cool.matrix(balance = False)
    # computing each chomosome copy number and median
    chromosomes_copy = [
        coverage(matrix, chrom, chromosomes)/(ploidy * cool.chromsizes[chrom])
        for chrom in chromosomes
    ]
    av_chrom_copy = np.nanmedian(chromosomes_copy)
                                
    # computing each contig coverage
    contigs_cov = {}
    contigs_copy = {}
    i = 0
    for contig in cool.chromnames:
        if contig in chromosomes:
            continue
        contigs_cov[contig] = coverage(matrix, contig, chromosomes)
        if contigs_cov[contig] == 0:
            continue
        i += 1
        if i % 100 == 0:
            print(f"{i} sequences processed.")
        contigs_copy[contig] = (contigs_cov[contig] / (ploidy * cool.chromsizes[contig])) / av_chrom_copy
    contigs_list = contigs_copy.keys()
    copy_df = pd.DataFrame([
        list(contigs_list),
        [cool.chromsizes[contig] for contig in contigs_list],
        [contigs_cov[contig] for contig in contigs_list], 
        [contigs_copy[contig] for contig in contigs_list]
    ]).T
    copy_df.columns = ["Contig", "Size", "Coverage", "Estimated # copy"]
    return copy_df
```

```python
library = "Tausonia_pullulans_contigs1"
binning = 100000
cool_file = f"/media/sardine/data_2/Eukaroytic_plasmids_manual_research/mcools/{library}.mcool"
plot_outpath = f"/media/sardine/data_2/Eukaroytic_plasmids_manual_research/contact_plots/{library}"
if not os.path.exists(plot_outpath):
    os.mkdir(plot_outpath)
cool = cooler.Cooler(f"{cool_file}::resolutions/{binning}")
matrix = cool.matrix(balance = False)
chromosomes = np.array([ chrom for chrom in list(cool.chromsizes.keys()) if chrom[:3] == "ENA" or (chrom[:3] != "ERR" and chrom[0] != "k" and chrom[:3] != "tig" and chrom != plasmid)])

for i in range(len(chromosomes)):
    chrom_tmp = chromosomes[1:] if i == 0 else np.concatenate([chromosomes[:i-1], chromosomes[i + 1:]]) if i < len(chromosomes) else chromsomes[:-1]
    cov = coverage(matrix, chromosomes[i], chrom_tmp)
    nb_copies = cov/(2 * cool.chromsizes[chromosomes[i]])
    print(f"Chromosome {chromosomes[i]}: {nb_copies} copies (coverage = {cov} reads).")

all_contigs_copy = compute_all_contigs_copy_number(cool, chromosomes, ploidy = 2)
```

```python
min_copy = 0.01
all_contigs_copy[all_contigs_copy["Estimated # copy"] >= min_copy]
```
